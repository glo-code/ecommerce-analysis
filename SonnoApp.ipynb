```python
df.drop(['User ID'],axis=1,inplace=True)
df
```

```python
df['Intervallo passi']=pd.cut(df['Daily Steps'],3,labels=['low','medium','high'])
df['Intervallo calorie']=pd.cut(df['Daily Steps'],3,labels=['low','medium','high'])
df['Age'].plot(kind='hist',bins=20)
df['Age'].skew(),df['Age'].kurt()
```

```python
df_add=df[['Age','Gender','Sleep Quality','Physical Activity Level','Dietary Habits','Sleep Disorders','Medication Usage','Hours','Intervallo passi','Intervallo calorie']]
df_add.tail(5)
```

```python
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder

encoder=OrdinalEncoder().set_output(transform='pandas')
encoder2=OneHotEncoder(sparse_output=False).set_output(transform='pandas')
physic=encoder.fit_transform(np.array(df_add['Physical Activity Level']).reshape(-1,1))
dietary=encoder.fit_transform(np.array(df_add['Dietary Habits']).reshape(-1,1))
passi=encoder.fit_transform(np.array(df_add['Intervallo passi']).reshape(-1,1))
calorie=encoder.fit_transform(np.array(df_add['Intervallo calorie']).reshape(-1,1))
sleep=encoder2.fit_transform(np.array(df_add['Sleep Disorders']).reshape(-1,1))
#sleep=pd.get_dummies(df_add['Sleep Disorders'])
medication=encoder2.fit_transform(np.array(df_add['Medication Usage']).reshape(-1,1))
sesso=encoder2.fit_transform(np.array(df_add['Gender']).reshape(-1,1))

#new_df=df_add.values.flatten().T
new_df=pd.concat([df['Age'],sesso,df['Sleep Quality'],physic,dietary,sleep,medication,df['Hours'],passi,calorie],axis=1)
new_df=new_df.astype(int)
new_df.columns=['Age','M','F','Sleep Quality','attività fisica','abitudini alimentari','disorder_yes','disorder_no','medication_no','medication_yes','Hours','passi','calorie']
corr=new_df.corr()
sns.heatmap(corr,annot=True)
corr
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler


y=new_df[['disorder_yes']]
X=new_df[['Age','Sleep Quality','abitudini alimentari','attività fisica','medication_yes','M','F','Hours']]

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
scaler=MinMaxScaler()
#Age=scaler.fit_transform(np.array(X_train['Age']).reshape(-1,1))
#X_train['Age']=Age
X_train
```

Machine Learning

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,recall_score,precision_score,f1_score,roc_auc_score,roc_curve,RocCurveDisplay,ConfusionMatrixDisplay,accuracy_score

logistic=LogisticRegression()
logistic.fit(X_train,y_train)
y_pred=logistic.predict(X_test)

y_p=pd.Series(y_pred)
print(y_p.value_counts())

confusion_matrix(y_test,y_pred)
precision_score(y_test,y_pred)
recall_score(y_test,y_pred)
f1_score(y_test,y_pred)
#RocCurveDisplay.from_predictions(y_test,y_pred)
```
